#### Formatting Strings with f-strings
An <b>f-string</b> (formatted string) lets you put variables directly inside text: start the string with `f` and put each variable in curly braces `{}`. You will see f-strings in most current Python and geospatial code.

```python
city = "Champaign"
num_points = 25
print(f"The {city} dataset has {num_points} points.")
```

You can also format numbers inside the braces. For example, `:.2f` rounds to two decimal places, which is handy for coordinates and measurements:

```python
lat = 40.116421
lon = -88.243385
print(f"Location: ({lat:.2f}, {lon:.2f})")
print(f"Area: {1234.5678:,.1f} square meters")
```

### Whitespace

<b>Blank lines</b> between code lines are ignored, so you can use blank lines to group things for ease of reading.

```python
a = 1
b = 4

c = 'cat'
```

White space <b>within</b> a line is often ignored, so you can condense, align, or spread out code for clarity.

```python
# these are all equivalent

a=3#comment
a = 3 # comment
a    =    3       #      comment
```

However, space is needed after or between <b>key words</b> and <b>commands</b> so Python can parse them properly.

White space <b>in the front of</b> a line is <b>very important</b> for python to work correctly. When typing code in a cell, make sure each regular line starts at the very beginning of the line with no leading space.

```python
# this works

a = 2
b = 3
```

```python
# this will fail

a = 2
 b = 3
```

```python
# this also fails

 a = 2
 b = 3
```

<b>Indentation</b>, typically using <b>Tab</b>, represents a group of lines (called a <b>code block</b>) that are related to the last unindented line. Each indented line in a cell needs to match up with all the other lines that are indented the same amount of times and the same amount of space (again usually in increments of Tab) needs to occur before each indented line. Although you can indent code blocks with spaces instead of Tabs, it is often easier to use Tab (and you need to be consistent throughout a script or within a cell).

```python
# example of indented block

a = 2
if a:
    print('a exists')
```

```python
# you can have multiple indentation levels

a = 2
if a:
    print('a exists')
b = 3
if b:
    print('b exists')
if a:
    if b:
        print('a and b exist')
```

```python
# indent with spaces
# this works, but Jupyter notebook will highlight keywords in red because it expects Tab
a = 2
if a:
  print('a exists')
b = 3
if b:
  print ('b exists')
```

```python
# this works but is --NOT-- recommended - make sure your indents match!

a = 2
if a:
  print ('a exists')
b = 3
if b:
    print ('b exists')
```

```python
# this works
# indentation is consistent within a code block

a = 2
if a:
  print ('a exists')
  print ('not sure if b exists')
```

```python
# this doesn't work
# indentation is not consistent within a code block

a = 2
if a:
  print ('a exists')
    print ('not sure if b exists')
```

#### Looping with enumerate and zip
`enumerate` gives you a counter along with each item, so you don't need `range(len(...))`. `zip` lets you loop over two (or more) lists at the same time. Together they cover a very common pattern in spatial code: matching coordinates with their attributes.

```python
cities = ["Chicago", "Champaign", "Springfield"]

for i, city in enumerate(cities):
    print(i, city)
```

```python
lats = [41.88, 40.12, 39.78]
lons = [-87.63, -88.24, -89.65]

for city, lat, lon in zip(cities, lats, lons):
    print(f"{city}: ({lat}, {lon})")
```

`zip` also makes it easy to build a list of coordinate pairs. Here they are in (longitude, latitude) order, the (x, y) order most geospatial libraries use:

```python
points = list(zip(lons, lats))
points
```

#### Keywords Affecting Loops

Certain keywords can affect how the loop functions:

```python
# stop if 7 is reached

for i in range(10):
    if i == 7:
        break
    print (i, end=" ")
```

```python
# prints '- no break' if loop completed without break

for i in range(10):
    if i == 12:
        break
    print (i, end=" ")
else:
    print ('- no break')
```

```python
# skips even numbers, but continues through loop after

for i in range(10):
    if i % 2 == 0:
        continue
    print (i, end=" ")
```

#### Functions: Swapping

```python
# Value changed within the function does not get reflected outside
a = 5
b = 10
def swap(a, b):
    temp = a
    a = b
    b = temp
swap(a, b)
print(a, b)
```

```python
# Value changed within the function is returned and assigned
a = 5
b = 10
def swap(a, b):
    temp = a
    a = b
    b = temp
    return a,b
(a,b)=swap(a, b)
print(a, b)
```

```python
def rescale(input_array):
    L=numpy.min(input_array)
    H=numpy.max(input_array)
    output_array=(input_array-L)/(H-L)
    return output_array

array = [0,1,2,4,5,10]
output=rescale(array)
print(output)
```

### Geospatial Data as Dictionaries: GeoJSON
<b>GeoJSON</b> is a common file format for geographic data. When Python reads it with the `json` library, you get ordinary <b>dictionaries</b> and <b>lists</b>, so everything you've learned in this notebook applies. The next cell creates a small GeoJSON file with two points.

```python
import json

geojson_text = '''{
  "type": "FeatureCollection",
  "features": [
    {"type": "Feature",
     "properties": {"name": "Champaign", "county": "Champaign"},
     "geometry": {"type": "Point", "coordinates": [-88.24, 40.12]}},
    {"type": "Feature",
     "properties": {"name": "Urbana", "county": "Champaign"},
     "geometry": {"type": "Point", "coordinates": [-88.21, 40.11]}}
  ]
}'''

# save it to a file, as you would with real data
with open('cities.geojson', 'w') as f:
    f.write(geojson_text)
```

```python
# read the file back in with the json library
with open('cities.geojson') as f:
    data = json.load(f)

print(type(data))
print(data['type'])
print(len(data['features']), 'features')
```

Each feature is a dictionary with <b>properties</b> (its attributes) and a <b>geometry</b> (its shape and coordinates). Notice that GeoJSON stores coordinates as <b>[longitude, latitude]</b>, that is, (x, y) order.

```python
for feature in data['features']:
    name = feature['properties']['name']
    lon, lat = feature['geometry']['coordinates']
    print(f"{name}: longitude {lon}, latitude {lat}")
```

#### Representing Rasters Using Lists

#### Representing Rasters Using Nested Lists

```python
raster=[[1,2,1,2],
[3,4,3,4],
[1,2,1,2],
[3,4,3,4]]
print("raster[3][0]=",raster[3][0])
print("raster[3][1]=",raster[3][1])
print("raster[3][2]=",raster[3][2])
print("raster[3][3]=",raster[3][3])
```

#### Printing Rasters Values

```python
for row in raster:
    print("row=",row)
```

```python
for row in raster:
    for element in row:
        print("element=",element)
```

```python
for r in range(0,4):
    for c in range(0,4):
        print("raster[",r,"][",c,"]=",raster[r][c])
```

```python
for r in range(0,len(raster)):
    for c in range(0,len(raster[0])):
        print("raster[",r,"][",c,"]=",raster[r][c])
```

#### From Nested Lists to NumPy Arrays
Raster libraries such as Rasterio (Module 3) give you raster data as a <b>NumPy array</b> rather than nested lists. An array works like the nested list above, but you can pick out rows, columns, or blocks with <b>slicing</b>, and compute statistics in one step.

```python
import numpy as np

grid = np.array(raster)
print(grid)
print("shape:", grid.shape)   # (rows, columns)
```

```python
print("row at index 3:", grid[3])      # one row
print("column at index 0:", grid[:, 0])   # one column
print("top-left 2x2 block:")
print(grid[:2, :2])
```

```python
print("mean value:", grid.mean())
print("max value:", grid.max())
print("cells with value 4:", (grid == 4).sum())
```
